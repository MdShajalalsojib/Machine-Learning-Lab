```python
import numpy as np
import pandas as pdp
```

```python
so = pdp.read_csv('bangla_spam.csv')
```

```python
so.head()
```

```python
so.head(10)
```

```python
import re

def clean_text(text):
    text = re.sub(r'[^\u0980-\u09FF\s]', '', str(text))
    return text

so['clean_text'] = so['text'].apply(clean_text)

so['tokens'] = so['clean_text'].apply(lambda x: x.split())

stopwords = ['আমি','তুমি','এই','ওই','সে','এটা']
so['tokens'] = so['tokens'].apply(lambda x: [word for word in x if word not in stopwords])

def stemming(words):
    return [word[:-1] if len(word) > 4 else word for word in words]

so['final'] = so['tokens'].apply(stemming)

so['final_text'] = so['final'].apply(lambda x: " ".join(x))
```

```python
so
```

## TF--IDF (Multaple word to number convert)

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer()
X = tfidf.fit_transform(so['final_text'])
```

## TF--IDF (word to number convert)

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
y = le.fit_transform(so['type'])
```

```python
from sklearn.model_selection import train_test_split

x = so['final_text']
y = so['type']

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.2)
```

## Model Apply

```python
from sklearn.naive_bayes import MultinomialNB

model = MultinomialNB()
model.fit(X_train, y_train)
```

## Score

```python
from sklearn.metrics import accuracy_score

y_pred = model.predict(X_test)

accuracy = accuracy_score(y_test, y_pred)
print("Accuracy:", accuracy)
```
