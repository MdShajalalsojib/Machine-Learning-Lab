```python
# apply function to a column
def fun(value):
    # print(type(value))
    if value<18:
        return 'child'
    else:
        return 'adult'

df['Age_description'] = df['Age'].apply(fun)
# df.drop('Name',axis=1) # This line is causing the error because 'Name' column was already dropped
```

```python
# !pip install scikit-learn
# Importing necessary libraries for encoding categorical variables
from sklearn.preprocessing import LabelEncoder, OrdinalEncoder, OneHotEncoder
```

```python
# label encoding
le = LabelEncoder()
df['Sex'] = le.fit_transform(df['Sex'])
```

```python
# one-hot encoding for age_description it removes the bias of the age_description column
ohe = OneHotEncoder()
encoded = ohe.fit_transform(df[['Age_description']])
# encoded.toarray()
```

```python
# to get the feature names after one-hot encoding
ohe.get_feature_names_out()
```

```python
# Create a new DataFrame with the one-hot encoded features
df_new = pd.DataFrame(
encoded.toarray(),columns=ohe.get_feature_names_out()
)
```

```python
df_new
```

```python
# x - x_min() / (x_max - x_min())
# min-max normalization
df['SibSp'] = (df['SibSp'] - df['SibSp'].min())/(df['SibSp'].max() - df['SibSp'].min())
```
