Feature Extraction

```python
feature_extraction = TfidfVectorizer(stop_words='english')

#Run the algorithm

feature_extraction.fit(x_train)

# Transform x_train and x_test
x_train_scaled =  feature_extraction.transform(x_train)
x_test_scaled =  feature_extraction.transform(x_test)
```
