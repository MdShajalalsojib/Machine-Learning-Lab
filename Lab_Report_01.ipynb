**Data cleaningt**

```python
import numpy as np
import pandas as pd
```

```python
sojib = pd.read_csv('bangla_spam.csv')
```

```python
sojib.head()
```

```python
sojib.dropna(inplace=True)
```

```python
import re

def clean_text(text):
    text = re.sub(r'[^\u0980-\u09FF\s]', '', str(text))
    return text

sojib['clean_text'] = sojib['text'].apply(clean_text)

sojib[['text','clean_text']].head()
```

```python
sojib['tokens'] = sojib['clean_text'].apply(lambda x: x.split())

sojib[['clean_text','tokens']].head()
```

```python
stopwords = ['আমি','তুমি','এই','ওই','সে','এটা']

sojib['tokens'] = sojib['tokens'].apply(lambda x: [word for word in x if word not in stopwords])

sojib[['tokens']].head()
```

```python
def stemming(words):
    return [word[:-1] if len(word) > 4 else word for word in words]

sojib['final'] = sojib['tokens'].apply(stemming)

sojib[['final']].head()
```

```python
sojib.to_csv("cleaned_dataset.csv", index=False)
```
